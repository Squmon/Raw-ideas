# Lorenz Attractor

```python
import numpy as np
from scipy.integrate import solve_ivp
import plotly.graph_objects as go
```

```python
np.random.seed(1234567890)
```

```python
# Define the Lorenz System
def lorenz_system(t, state, sigma, beta, rho):
    xi, yi, zi, x, y, z = state
    dxdt = sigma * (y - x)
    dydt = x * (rho - z) - y
    dzdt = x * y - beta * z

    dxidt = x
    dyidt = y
    dzidt = z - 40
    return [dxidt, dyidt, dzidt, dxdt, dydt, dzdt]
```

```python
# Solve the Lorenz System
# Parameters
sigma = 10.0
beta = 8.0 / 3.0
rho = 28.0

# Initial conditions
initial_state = [1., 0., 0., 1.0, 1.0, 1.0]

# Time span
t_span = (0, 50)
t_eval = np.linspace(t_span[0], t_span[1], 10000)
```

```python
solution = solve_ivp(lorenz_system, t_span, initial_state, t_eval=t_eval, args=(sigma, beta, rho))
```

```python
def show_all(solution):
    xi, yi, zi, x, y, z = solution.y
    time = solution.t

    # Visualize the Lorenz Attractor in 3D
    fig = go.Figure()
    fig.add_trace(go.Scatter3d(x=x, y=y, z=z, mode='lines', line=dict(color=time, width=2), ))
    fig.update_layout(
        title='Lorenz Attractor',
        scene=dict(
            xaxis_title='X',
            yaxis_title='Y',
            zaxis_title='Z'
        ),
        template='plotly_dark',
    )

    fig.show()

    # Visualize integrated Lorenz Attractor in 3D
    fig = go.Figure()
    fig.add_trace(go.Scatter3d(x=xi, y=yi, z=zi, mode='lines', line=dict(color=time, width=2)))

    fig.update_layout(
        title='Lorenz Attractor',
        scene=dict(
            xaxis_title='X',
            yaxis_title='Y',
            zaxis_title='Z'
        ),
        template='plotly_dark',
    )

    fig.show()
```

```python
show_all(solution)
```

```python
# Define the Lorenz System

W = np.random.randn(3, 3, 3)
W = np.transpose(W, (1, 0, 2)) - W
W *= 0.1
def lorenz_system(t, state, sigma, beta, rho):
    xi, yi, zi, x, y, z = state
    dxdt = sigma * (y - x)
    dydt = x * (rho - z) - y
    dzdt = x * y - beta * z
    G = np.einsum('j, k, ijk -> i', state[:3], state[3:], W)


    dxidt = G[0]
    dyidt = G[1]
    dzidt = G[2]
    return [dxidt, dyidt, dzidt, dxdt, dydt, dzdt]
```

```python
solution = solve_ivp(lorenz_system, t_span, initial_state, t_eval=t_eval, args=(sigma, beta, rho))
show_all(solution)
```
